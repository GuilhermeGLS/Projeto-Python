# ELIF
- if condicao:
   > retorna se codicao if verdadeiro
- elif condicao2:
   > retorna se condicao falsa e condicao2 verdaeira
- else:
    >retorna se condicao 1 e 2 falsa

```python
meta = 20000
bonus1 = 0.03
bonus2 = 0.07

venda = 100000

if venda < meta:
    print(f"Não teve comissão")
elif venda >= (meta * 2):
    comissao = venda*bonus2
    print(f"parabéns bateu o dobro da meta sua comissão foi {comissao}")
else:
    comissao = venda*bonus1
    print(f"sua comissão foi {comissao}")
```

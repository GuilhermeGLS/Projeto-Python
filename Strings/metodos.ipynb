# Metodos
Normalmente se utiliza nessa estrutura .método(params) os params podem ser vazios também.

- .capitalize() -> coloca a 1° letra em Maiúsculo

```python
texto = "guilherme"
print(texto.capitalize())
```

- .casefold() -> Tranforma todas as letrar em minusculo. (Tamém existe o .lower())

```python
texto = "GUILHERME"
print(texto.casefold())
print(texto.lower())
```

- .count() ->  Conta a quantidade de vezes que um valor aparece na string

```python
texto = "guilhermelima@opcaodistribuidora.com.br"
print(texto.count('a'))
```

- .endswith() -> Verifica se o texto termina com um valor especifico. Retorna um boolean true or false

```python
texto = "guilhermelima@gmail.com"
print(texto.endswith('gmail.com'))
```

- .find() Procura um texto dentro de outro texto, e retorna a posição do texto encontrado

```python
texto = "guilhermelima@gmail.com"
print(texto.find('@'))
```

- .isalnum() -> Verifica se um texto é todo feito com caracteres alfanumericos (letras e números) -> letras com acento ou ç são considerados letras. Retornando true or false

```python
texto = 'guilherme123'
print(texto.isalnum())
```

- .isalpha() -> verifica se é todo feito de letras

```python
texto = 'guilherme'
print(texto.isalpha())
```

- .isnumeric() -> verifica se é todo feito de números

```python
texto = '610.773.663-62'
#utilizando o replace para ser true no .isnumeric()
print(texto.replace('.', '').replace('-', '').isnumeric())
```

- .replace -> troca um texto por outro 

```python
texto = '1000.00'
print(texto.replace('.', ','))
```

- .split() -> Separa uma string de acordo com um delemitador em vários textos diferentes

```python
texto = "guilhermelima@gmail.com"
print(texto.split('@'))
```

- .splitlines -> separa um texto em vários textos de acordo com a "entrada" do texto

```python
faturamento = 1200
texto = f'''
Olá, Bom dia
Venho por meio deste e-mail informar o faturamento da loja no dia de hoje
faturamento = {faturamento}
'''

print(texto.splitlines())
```

- .starwith -> Verifica se a string começa com determinado texto.

```python
texto = "OPCAOpromotor12345"

print(texto.startswith('OPCAO'))
```

- .strip() -> Retira caracteres indesejados dos textos. Por padrão, retira espaços extras no inicio e no final

```python
texto = "    OPCAOpromotor12345     "
print(texto.strip())
```

- .title() -> Coloca a 1° Letra de cada palavra maiuscula

```python
texto = "guilherme lima e silva"

print(texto.title())
```

- .upper() -> Coloca todos os caracteres maiusculos

```python
texto = "guilherme lima e silva"

print(texto.upper())
```

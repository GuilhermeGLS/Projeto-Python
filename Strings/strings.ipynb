# Indice Negativo

```python
texto = 'guilherme'
email = "guilhermelimaesilva.p@gmail.com"

print(email[-1])


# Pegando pedações de texto .Split por exemplo.
# Como são todos objetos podemos cortar e pegar pedaçoes de strings usando os (:)
# Muda de acordo se o indice está antes ou depois dos 2 pontos (:)

print(email[:-10])
print(email[-10:])
```

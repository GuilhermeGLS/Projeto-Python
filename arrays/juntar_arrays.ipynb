##  Juntar listas, Ordenar, Cuidados especiais

### 2 Formas

- lista1.extend(lista2)
- lista_nova = lista1 + lista2

OBS: O método .append() não junta listas, apenas adiciona um valor no final da lista

```python
# Primeira opção
produtos = ["apple tv", "mac", "iphone x", "ipad", "apple watch", "macbook", "airpods"]
novos_produtos = ["iphone17", "ioculos"]

produtos.extend(novos_produtos)

print(produtos)
```

```python
# Segunda opção

produtos = ["apple tv", "mac", "iphone x", "ipad", "apple watch", "macbook", "airpods"]
novos_produtos = ["iphone17", "ioculos"]

todos_produtos = produtos + novos_produtos

print(todos_produtos)
```

### Ordenar listas

```python
# Listas.sort()
produtos = ["apple tv", "mac", "iphone x", "ipad", "apple watch", "macbook", "airpods"]
produtos.sort(reverse=True)
print(produtos)
```

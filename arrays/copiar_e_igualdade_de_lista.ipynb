# Copiar e 'igualdade' de listas

Utilizar o metodo .copy()

```python
lista = ['iphone', 'ipad', 'ipod']
lista2 = lista.copy()

lista2[1] =  'iphone11'

print(lista)
print(lista2)




```

```python
#Pode também utilizar o (:)

lista = ['iphone', 'ipad', 'ipod']
lista2 = lista[:]

lista2[1] =  'iphone11'

print(lista)
print(lista2)
```

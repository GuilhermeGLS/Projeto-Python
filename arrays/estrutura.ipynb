# Descobrindo o índice da lista

- Utilizando o método index
índice = lista.index('item')

```python
produtos = ["tv", "celular", "tablet", "mouse", "teclado", "geladeira", "forno"]
estoque = [100, 150, 100, 120, 70, 90, 80]


#descobrindo a quantidade estoque produto geladeira sem contar
# teste trocar os produtos em produtos.index('troque aqui')

indice_geladeira = produtos.index('celular')
qtd_estoque = estoque[indice_geladeira]
print(f"{produtos[indice_geladeira]} possui {qtd_estoque} unidades")
```

```python
# sistema onde a pessoa digita o nome do produto e retornar a qtd no estoque, caso não tenha o produto retorna que produto não existe na lista

produtos = ["tv", "celular", "tablet", "mouse", "teclado", "geladeira", "forno"]
estoque = [100, 150, 100, 120, 70, 90, 80]

pesquisa = input('Digite um produto').lower()

if pesquisa in produtos:
    indice_geladeira = produtos.index(pesquisa)
    qtd_estoque = estoque[indice_geladeira]
    print(f"{produtos[indice_geladeira]} possui {qtd_estoque} unidades")
else:
    print(f"Não possuimos o produto {pesquisa}")
```

## Lista de Listas

```python
# Criando a estrutura de vendas
vendas = {
    'Guilherme': [100, 200],
    'Helano': [200, 500],
    'Mateus': [150, 300, 400],
    'Jardel': [200, 150, 90]
}

# 1. Acessando as vendas de um vendedor específico:
print(vendas['Guilherme'])
# Saída: [100, 200]

# 2. Adicionando uma nova venda para o Guilherme:
vendas['Guilherme'].append(150)
print(vendas['Guilherme'])
# Saída: [100, 200, 150]

# 3. Calculando o total de vendas de cada um:
for vendedor, lista_vendas in vendas.items():
    total = sum(lista_vendas)
    print(f"{vendedor} vendeu um total de R$ {total}")
```

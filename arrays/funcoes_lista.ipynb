## Maior e Menor valor
- maior  = max(lista)
- menor = min(lista)

```python
vendas = [1000, 1500, 10000, 120, 700, 900, 100]

maior_venda = max(vendas)
menor_venda = min(vendas)
i = vendas.index(menor_venda)

print(f"A maior venda foi {maior_venda} a menor venda foi {menor_venda} menor ele se localiza {i}")
```

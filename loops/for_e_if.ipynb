```python
vendas = [25000, 29000, 22200, 8000, 15870, 19900, 50000, 5000]
meta = 10000
qtd_bateu_meta = 0

for venda in vendas:
    if venda >= meta:
        qtd_bateu_meta +=1
        
print(f'{qtd_bateu_meta} de {len(vendas)} pessoas atingiu a meta')  
print(f'o percentual foi {qtd_bateu_meta/len(vendas):.1%}')  
```

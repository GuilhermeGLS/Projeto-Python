### For item in lista

```python
produtos = ['coca', 'pepsi', 'guarana', 'sprite', 'fanta']
precos = [15000, 12000, 13000, 5000, 250]
```

```python
#preco com imposto

for preco in precos:
    print(preco * 1.1)
```

### For i in range

```python
produtos = ['coca', 'pepsi', 'guarana', 'sprite', 'fanta']
precos = [15000, 12000, 13000, 5000, 250]

for i in range(len(precos)):
    produto = produtos[i]
    preco = precos[i]
    print(produto, preco)
```

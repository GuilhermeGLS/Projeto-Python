# Método Format


```python
faturamento = 2000
custo = 500
lucro = faturamento - custo

print("faturamento é {} o custo foi {} e o lucro deu {}".format(faturamento, custo, lucro))
```

# Com F-string

```python
faturamento = 23000
custo = 10000
lucro = faturamento - custo

print(f"O Faturamento foi {faturamento} e o lucro foi de {lucro}")
```

# Mudando o Tipo da variável
